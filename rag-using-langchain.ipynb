```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
# from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

## Step 1a - Indexing (Document Ingestion)

```python
video_id = "Gfr50f6ZBvo" # only the ID, not full URL LPZh9BOjkQs Gfr50f6ZBvo  VwW6ItSjkzY
try:
    # If you don’t care which language, this returns the “best” one
    # transcript_list = YouTubeTranscriptApi.get_transcript(video_id, languages=["en"])
    ytt_api = YouTubeTranscriptApi()
    transcript_list = ytt_api.fetch(video_id, languages=["en"])

    # Flatten it to plain text
    # transcript = " ".join(chunk["text"] for chunk in transcript_list)
    transcript = " ".join(chunk.text for chunk in transcript_list)
    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
transcript_list
```

## Step 1b - Indexing (Text Splitting)

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[100]
```

## Step 1c & 1d - Indexing (Embedding Generation and Storing in Vector Store)

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['76e159c8-3313-4e04-9064-b07c7da7b72a'])
```

## Step 2 - Retrieval

```python
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 4})
```

```python
retriever
```

```python
retriever.invoke('What is deepmind')
```

## Step 3 - Augmentation

```python
# llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.2)
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",      # or "gemini-2.0-flash", "gemini-1.5-pro" etc.
    temperature=0.7,
    google_api_key=""
)
```

```python
prompt = PromptTemplate(
    template="""
      You are a helpful assistant.
      Answer ONLY from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables = ['context', 'question']
)
```

```python
question          = "is the topic of nuclear fusion discussed in this video? if yes then what was discussed"
retrieved_docs    = retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
context_text
```

```python
final_prompt = prompt.invoke({"context": context_text, "question": question})
```

```python
final_prompt
```

## Step 4 - Generation

```python
answer = llm.invoke(final_prompt)
print(answer.content)
# print(answer)
```

## Building a Chain

```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke('who is Demis')
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('Can you summarize the video')
```
